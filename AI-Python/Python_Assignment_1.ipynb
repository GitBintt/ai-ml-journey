# Answer to the Question Number : 1

```python
inp = input()
numbers = inp.split()
result = []

for num in numbers:
  if int(num) % 4 == 0 and int(num) % 8 != 0 :
    result.append(int(num))

print(result)
```

# Answer to the Question Number : 2

```python
inp = input()
numbers = inp.split()
x = float(numbers[0])
y = float(numbers[1])

def compute_ratio(x , y):
          try:
            ratio = (x + y) / (x * y)
            return ratio
          except ZeroDivisionError:
            return "Error: Division not possible"

print(compute_ratio(x , y))
```

# Answer to the Question Number : 3

```python
with open("./sample_data/data.txt", "r") as file:
  content = file.read()

  words = content.split()
  longest = words[0]

  for word in words:
    if len(word.strip()) > len(longest.strip()):
      longest = word

  print(longest.strip())
```

# Answer to the Question Number : 5

```python
inp = input()
message = inp.lower()
reverse = message[::-1]

if message == reverse:
  print("Palindrome")
else:
  print("Not Palindrome")
```
